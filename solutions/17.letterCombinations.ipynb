```python
from typing import List
from itertools import product


class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        
        str_digits = str(digits)
        if len(str_digits) == 0 or len(str_digits) == str_digits.count('1'):
            return []

        str_digit_clean = str_digits.replace('1', '')

        keyboard = {
            "2": ["a", "b", "c"],
            "3": ["d", "e", "f"],
            "4": ["g", "h", "i"],
            "5": ["j", "k", "l"],
            "6": ["m", "n", "o"],
            "7": ["p", "q", "r", "s"],
            "8": ["t", "u", "v"],
            "9": ["w", "x", "y", "z"]
        }

        if len(str_digit_clean) == 1:
            return keyboard[str_digit_clean[0]]

        letter_groups = [keyboard[d] for d in str_digit_clean]

        all_combinations = ["".join(tup) for tup in product(*letter_groups)]

        return all_combinations


solution = Solution()

print(solution.letterCombinations(digits="43"))
print(solution.letterCombinations(digits=""))
print(solution.letterCombinations(digits="2"))
print(solution.letterCombinations(digits="1111"))
print(solution.letterCombinations(digits="12231"))
```
