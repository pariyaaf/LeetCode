```python
class Solution:
    def generate(self, numRows: int) -> List[List[int]]:
        final_triangle = []
        for i in range(numRows):
            if i == 0:
                final_triangle.append([1])
            else:
                final_triangle.append([])
                for j in range(i+1):
                    len_of_last_row = len(final_triangle[i-1])
                    sum_value = 0
                    if j != 0:
                        sum_value +=final_triangle[i-1][j-1]
                    if j != len_of_last_row:
                        sum_value +=final_triangle[i-1][j]
                    final_triangle[i].append(sum_value)
        return final_triangle
```
