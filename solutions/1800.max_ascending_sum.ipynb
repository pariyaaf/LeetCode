```python
from typing import List
class Solution:
    def maxAscendingSum(self, nums: List[int]) -> int:
        all_substring = [nums[0]]
        this_substring = nums[0]

        for i in range(len(nums)-1):

            if nums[i] < nums[i+1]:
                this_substring += nums[i+1]
            if nums[i] >= nums[i+1] or i == len(nums)-2:
                all_substring.append(this_substring)
                this_substring = nums[i+1]
        return max(all_substring)
        


solution = Solution()

print(solution.maxAscendingSum([6]))

print(solution.maxAscendingSum([100,10,1]))

print(solution.maxAscendingSum([10,20,30,5,10,50]))
print(solution.maxAscendingSum([10,20,30,40,50]) )
print(solution.maxAscendingSum([12,17,15,13,10,11,12]))  
```
