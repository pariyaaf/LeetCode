```python
from typing import List
class Solution:
    def commonChars(self, words: List[str]) -> List[str]:
        same_letters = []
        
        for letter in words[0]:
            counter = 0
            
            for i in range(1, len(words)):
                word = words[i]  
                if letter in word:
                    words[i] = word.replace(letter, '_', 1) 
                    counter += 1
            
            if counter == len(words) - 1:
                same_letters.append(letter)
        
        return same_letters
```
