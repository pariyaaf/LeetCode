```python
class Solution:
    def isUgly(self, n: int) -> bool:
        if n == 1:
            return True
        if n == 0:
            return False

        for num in [2, 3 ,5]:
            while n % num == 0:
                n //= num
                
        return n == 1

s = Solution()
print(s.isUgly(8))
print(s.isUgly(2))
print(s.isUgly(4))
print(s.isUgly(10))
print(s.isUgly(7))
print(s.isUgly(14))
```
