```python
class Solution:
    def intToRoman(self, num: int) -> str:
        roman = {1000: 'M', 500: 'D', 100: 'C', 50: 'L', 10: 'X', 5: 'V', 1: 'I'}
        final_roman = []
### حتما درست کن حروف و اینا رو اوکی کن حتما
        counter = 1
        for i in str(num)[::-1]:
            this_roman = ""
            real_num = int(i) * counter
            largest_digit = int(max(str(real_num)))
            if largest_digit in [1,2,3]:
                this_roman += roman[int(real_num/ largest_digit)] * largest_digit
                
            if largest_digit in [5,6,7,8]:
                base = 5 * counter
                this_roman += roman[base]
                another_counter = int(max(str(real_num - base)))
                this_roman += roman[counter] * another_counter

            
            elif largest_digit in [4,9]:
                this_roman += roman[counter]
                this_roman += roman[real_num+counter] # ok
                
            counter = counter * 10
            final_roman.insert(0, this_roman)

        return  "".join(final_roman)

            
            
            
            
            



solution = Solution()

print(solution.intToRoman(3749))  
print(solution.intToRoman(58)) 
print(solution.intToRoman(1994))      
print(solution.intToRoman(444))      

```

```python
# faster
class Solution:
  def intToRoman(self, num: int) -> str:
    valueSymbols = [(1000, 'M'), (900, 'CM'),
                    (500, 'D'), (400, 'CD'),
                    (100, 'C'), (90, 'XC'),
                    (50, 'L'), (40, 'XL'),
                    (10, 'X'), (9, 'IX'),
                    (5, 'V'), (4, 'IV'),
                    (1, 'I')]
    ans = []

    for value, symbol in valueSymbols:
      if num == 0:
        break
      count, num = divmod(num, value)
      ans.append(symbol * count)

    return ''.join(ans)
```
