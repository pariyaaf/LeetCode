```python
class Solution:
    def fib(self, n: int) -> int:
        if n == 0:
            return 0
        elif n == 1:
            return 1
        else:
            return self.fib(n-1) + self.fib(n-2)  
```

```python
class Solution:
    def fib(self, n: int) -> int:
        memo = {0: 0, 1: 1}  
        def helper(n):
            if n not in memo:
                memo[n] = helper(n-1) + helper(n-2)
            return memo[n]
        
        return helper(n)
```
