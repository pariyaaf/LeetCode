```python
class Solution:
    def mySqrt(self, x: int) -> int:
        if x == 0:
            return 0

        if x == 1:
            return 1
            
        n = x 
        y = n >> 1
        while y < n:
            n = y
            y = (n + x // n) >> 1
        return n


s = Solution()
print(s.mySqrt(1))
print(s.mySqrt(4))
print(s.mySqrt(9))
```

```python
class Solution:
    def mySqrt(self, x: int) -> int:
        if x == 0:
            return 0

        if x == 1:
            return 1
            
        n = x # 8
        y = n >> 1 # 4
        
        while y < n: #4<8  -   3<4   - 2<3   - 3<2 تامام
            n = y # 4      -    3      2
            y = (n + x // n) >> 1 # (4+ 2)>>1 =>3  -   (5)>>1 =>2   -   (6)>>1 => 3
        return n
#پس ان میشه همون ایگیرگ حساب شده در ترم قبلی

s = Solution()
print(s.mySqrt(8))
```

```python
def sqrt(n):
    if n == 0:
        return 0
    x = n
    y = n >> 1  # مقدار اولیه را نصف مقدار اصلی قرار می‌دهیم
    while y < x:
        x = y
        y = (x + n // x) >> 1  # روش نیوتن با شیفت راست
    return x

print(sqrt(16))  # خروجی: 4
print(sqrt(25))  # خروجی: 5
print(sqrt(8)) # خروجی: 10
```
