```python
from typing import List

class Solution:
        
    def maxArea(self, height: List[int]) -> int:
        max_area = 0
        right = len(height) - 1
        left = 0
        
        while True:
            first = height[left]
            second = height[right]
            area = (right - left) * min(first, second)
            
            max_area = max(max_area, area)
                
            if first < second:
                left += 1
            else:
                right -= 1
                
            if right == left:
                break
        return max_area

s = Solution()

print(s.maxArea( height = [1,8,6,2,5,4,8,3,7]))

print(s.maxArea( height = [0,1]))
```
