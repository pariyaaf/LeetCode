```python
class Solution:
    def convert(self, s: str, numRows: int) -> str:
        if numRows == 1 or numRows >= len(s):  
            return s
        final_str = [""] * numRows 
        formula = 2 * numRows - 2  

        for i in range(numRows):
            x = i
            while x < len(s):
                final_str[i] += s[x]
                if 0 < i < numRows - 1:  
                    mid_index = x + formula - 2 * i
                    if mid_index < len(s):
                        final_str[i] += s[mid_index]
                x += formula 

        return "".join(final_str)  

s = Solution()
print(s.convert( s = "PAYPALISHIRING", numRows = 4))
print('PINALSIGYAHRPI')
print( s.convert(s = "PAYPALISHIRING", numRows = 3))
print('PAHNAPLSIIGYIR')
```
