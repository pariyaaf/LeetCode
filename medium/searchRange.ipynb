```python
from typing import List

class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        first_index = 0
        last_index = len(nums)-1
        all_indexs = []
        
        while first_index <= last_index:
            middle_index = (last_index + first_index) // 2
            middle_number = nums[middle_index]
            
            if nums[middle_index] == target:
                all_indexs.append(middle_index)
                
                left = middle_index - 1
                while left >= 0 and nums[left] == target:
                    all_indexs.append(left)
                    left -= 1
                
                right = middle_index + 1
                while right < len(nums) and nums[right] == target:
                    all_indexs.append(right)
                    right += 1
                
                break
            
            elif middle_number < target:
                first_index = middle_index + 1
            
            elif middle_number > target:
                last_index = middle_index - 1
        if all_indexs:
            return [min(all_indexs), max(all_indexs)]
        else:
            return [-1, -1]
            
        
s = Solution()
print(s.searchRange([3,3,3], target=3))
print(s.searchRange([], target = 8))
print(s.searchRange([5], target = 8))
print(s.searchRange([8], target = 8))

print(s.searchRange([5,7,7,8,8,10], target = 8))
print(s.searchRange([5,7,7,8,8,10], target = 6))
```
