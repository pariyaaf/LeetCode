```python
class Solution:
    def nthUglyNumber(self, n: int) -> int:
        ugly_numbers = [1]
        step_of_two, step_of_three, step_of_five = 0,0,0
        
        this_two = 2
        this_three = 3 
        this_five = 5
        
        while len(ugly_numbers) < n:
            new_ugly = min(this_five,this_three,this_two)
            ugly_numbers.append(new_ugly)

            if new_ugly == this_two:
                step_of_two += 1
                this_two = ugly_numbers[step_of_two] * 2
            
            if new_ugly == this_three:
                step_of_three += 1
                this_three = ugly_numbers[step_of_three] * 3
            
            if new_ugly == this_five:
                step_of_five += 1
                this_five = ugly_numbers[step_of_five] * 5


        return ugly_numbers[-1]
            
            
            
s = Solution()
print(s.nthUglyNumber(6))
print(s.nthUglyNumber(15))

print(s.nthUglyNumber(1))
```
