```python
# O(n2)

class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        if s == "":
            return 0
            
        if len(s) == s.count(s[0]):
            return 1
            
        if len(s) == s.count(s[0])-1:
            return 2

        all_substring = []

        for i in range(len(s)):
            new_substring = s[i]
            for j in range(i+1,len(s)):
                if s[j] in new_substring:
                    break
                
                new_substring = new_substring + s[j]
                all_substring.append(new_substring)

        return  len(max(all_substring, key=len))
                    
            



s = Solution()
print(s.lengthOfLongestSubstring("")) 

print(s.lengthOfLongestSubstring("abcabcbcdef"))  
print(s.lengthOfLongestSubstring("bbbbb"))   
```

```python
# O(n)
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        char_set = set()
        left = 0
        max_length = 0

        for right in range(len(s)):
            while s[right] in char_set:
                char_set.remove(s[left])
                left += 1  

            char_set.add(s[right])
            max_length = max(max_length, right - left + 1) 

        return max_length

s = Solution()
print(s.lengthOfLongestSubstring("abcabcbb")) 
print(s.lengthOfLongestSubstring("bbbbb")) 
print(s.lengthOfLongestSubstring("pwwkew"))
```
