```python
from typing import List
class Solution:
    def validStrings(self, n: int) -> List[str]:
        if n == 1:
            return ['0','1']
            
        max_num = (2 ** n) -1
        min_num =2 **(n-2) -1
        all_valid_substring = []
        for i in range(max_num, min_num , -1):
            this_num = bin(i)[2:]
            this_num = this_num.zfill(n) 
            if '00' not in this_num:
                all_valid_substring.append(this_num)
        return all_valid_substring
            

s = Solution()

print(s.validStrings(3))
print(s.validStrings(1))
print(s.validStrings(5))

```

```python
from typing import List

class Solution:
    def validStrings(self, n: int) -> List[str]:
        result = []
        
        def backtrack(current):
            if len(current) == n:
                result.append(current)
                return
            
            if not current or current[-1] == "1":
                backtrack(current + "0")
                backtrack(current + "1")
            else:
                backtrack(current + "1")
        
        backtrack("")
        
        return result

s = Solution()
print(s.validStrings(5))
```
