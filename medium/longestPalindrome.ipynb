```python
# bad 

class Solution:
    def longestPalindrome(self, s: str) -> str:
        max_length = len(s)
        palindromic_substrings = []
        for i in range(max_length):
            current_substring = s[i]
            palindromic_substrings.append(current_substring)
            for j in range(i + 1, max_length):
                new_char = s[j]
                new_substring = current_substring + new_char
                if new_substring == new_substring[::-1]:
                    palindromic_substrings.append(new_substring)
                current_substring = new_substring
        return max(palindromic_substrings, key=len)

                


s = Solution()
print(s.longestPalindrome(s = "babad"))
print(s.longestPalindrome(s = "cdbb"))
```

```python
# best
class Solution:
    def longestPalindrome(self, s: str) -> str:
        if not s:
            return ""

        transformed = "#" + "#".join(s) + "#"
        n = len(transformed)
        p = [0] * n
        center, right = 0, 0
        max_len, start = 0, 0
        for i in range(n):
            mirror = 2 * center - i 
            
            if i < right:
                p[i] = min(right - i, p[mirror])
            
            while i + p[i] + 1 < n and i - p[i] - 1 >= 0 and transformed[i + p[i] + 1] == transformed[i - p[i] - 1]:
                p[i] += 1
            
            if i + p[i] > right:
                center, right = i, i + p[i]

            if p[i] > max_len:
                max_len = p[i]
                start = (i - max_len) // 2
        
        return s[start:start + max_len]

s = Solution()
print(s.longestPalindrome("babad")) 
print(s.longestPalindrome("cdbb")) 
```
