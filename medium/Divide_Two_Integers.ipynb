```python
class Solution:
    def divide(self, dividend: int, divisor: int) -> int:
        counter = 0
        
        divisor_sign = -1 if divisor < 0 else 1
        dividend_sign = -1 if dividend < 0 else 1
        divisor, dividend = abs(divisor), abs(dividend)

        while dividend >= divisor:
            temp, multiple = divisor, 1
            while dividend >= (temp << 1):
                temp <<= 1
                multiple <<= 1
            dividend -= temp
            counter += multiple

        return max(-2**31, min((counter * divisor_sign * dividend_sign), 2**31 -1))

solution = Solution()
print(solution.divide(dividend = 2147483647, divisor = 2))  
print(solution.divide(dividend = 10, divisor = 3))  
print(solution.divide(dividend = 7, divisor = -3)) 
print(solution.divide(dividend = 3000, divisor = -3))     

```
