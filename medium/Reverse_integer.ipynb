```python
class Solution:
    def reverse(self, x: int) -> int:
        if x == 0:
            return 0
        
        sign = -1 if x < 0 else 1
        trimmed_str = str(abs(x)).lstrip('0')
        trimmed_reversed_str = trimmed_str[::-1]
        reversed_int = sign * int(trimmed_reversed_str)
  
        if reversed_int < -2** 31 or 2**31-1 < reversed_int:
            return 0
        else:
             return reversed_int



solution = Solution()
print(solution.reverse(0))  
print(solution.reverse(1534236469))  

print(solution.reverse(123))  
print(solution.reverse(-123))  
print(solution.reverse(120))  

print(solution.reverse(3749))  
print(solution.reverse(58)) 
print(solution.reverse(1994))      
print(solution.reverse(444))      
```
