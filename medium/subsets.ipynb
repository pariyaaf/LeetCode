```python
from typing import List


class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        subsets_list = [[]]

        for num in nums:
            new_subsets = [subset + [num] for subset in subsets_list]
            print(new_subsets, num)
            subsets_list.extend(new_subsets)

        return subsets_list


s = Solution()
print(s.subsets(nums = [1,2,3,4]))
print(s.subsets(nums = [0]))
```
