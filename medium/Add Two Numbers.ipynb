```python
from typing import List, Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        carry = 0
        dummy_head = ListNode(0) 
        current = dummy_head
        
        while l1 or l2 or carry:
            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0
            print(val1, val2)

            sum_val = val1 + val2 + carry
            carry = sum_val // 10 
            current.next = ListNode(sum_val % 10)
            current = current.next
            
            if l1: l1 = l1.next
            if l2: l2 = l2.next
        
        return dummy_head.next 

```

```python
# تابع کمکی برای چاپ Linked List
def print_linked_list(head: Optional[ListNode]) -> None:
    current = head
    while current:
        print(current.val, end=" -> ")
        current = current.next
    print("None")

# ساختن لیست پیوندی 1: [2, 4, 3] معادل عدد 342
l1 = ListNode(2)
l1.next = ListNode(4)
l1.next.next = ListNode(3)

# ساختن لیست پیوندی 2: [5, 6, 4] معادل عدد 465
l2 = ListNode(5)
l2.next = ListNode(6)
l2.next.next = ListNode(4)

solution = Solution()

# فراخوانی تابع و چاپ نتیجه
result = solution.addTwoNumbers(l1, l2)
print("Result: ", end="")
print_linked_list(result)

# ساختن مثال‌های دیگری از لیست پیوندی
# مثال 2: [0] + [0] -> [0]
l1 = ListNode(0)
l2 = ListNode(0)
result = solution.addTwoNumbers(l1, l2)
print("Result: ", end="")
print_linked_list(result)

# مثال 3: [9,9,9] + [1] -> [0, 0, 0, 1]
l1 = ListNode(9)
l1.next = ListNode(9)
l1.next.next = ListNode(9)
l2 = ListNode(1)
result = solution.addTwoNumbers(l1, l2)
print("Result: ", end="")
print_linked_list(result)
```
