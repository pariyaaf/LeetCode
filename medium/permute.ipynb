```python
class Solution:

    def permute(self, nums: List[int]) -> List[List[int]]:
        result = []
        def backtrack(selected, remaining):
            if not remaining:
                result.append(selected)
            for i in range(len(remaining)):
                backtrack(selected + [remaining[i]], remaining[:i]+ remaining[i+1:])
        backtrack([], nums)
        return result

s = Solution()
print(s.permute( nums = [1,2,3]))
```
