```python
from typing import List

class Solution:
    def longestMonotonicSubarray(self, nums: List[int]) -> int:
        longest = 1
        for i in range(0, len(nums)+1):
            for j in range(i+2, len(nums)+1):
                substring = nums[i:j]
                decreasing, increasing = 0, 0

                for index in range(0,len(substring)-1):
                    if substring[index] < substring[index+1]:
                        increasing += 1

                    if substring[index] > substring[index+1]:
                        decreasing += 1

                if decreasing == len(substring)-1 or increasing == len(substring)-1:
                    longest = max(longest, len(substring))

        return longest

solution = Solution()

print(solution.longestMonotonicSubarray(nums = [3, 2, 4], target = 6))  
print(solution.longestMonotonicSubarray([3, 3, 3, 3]))   
print(solution.longestMonotonicSubarray([3, 2, 1]))       
print(solution.longestMonotonicSubarray([1,2]))        
print(solution.longestMonotonicSubarray([40,1]))       
```
