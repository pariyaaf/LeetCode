```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        str_num = str(x)
        reverse_str_num = str_num[::-1]
        return reverse_str_num == str_num


solution = Solution()

print(solution.isPalindrome(x=121))  
print(solution.isPalindrome(x=10)) 
print(solution.isPalindrome(x=-121))      
      
```
