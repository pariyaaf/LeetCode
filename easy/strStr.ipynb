```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        return haystack.find(needle)

s = Solution()

print(s.strStr(haystack = "sadbutsad", needle = "sad"))
print(s.strStr(haystack = "sasad", needle = "sad"))

print(s.strStr(haystack = "leetcode", needle = "leeto"))
# print(s.strStr())
```
