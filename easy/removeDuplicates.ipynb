```python
from typing import List
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        k = 0
        for i in range(len(nums)):
            current_num = nums[i]
            if current_num == '_' or nums.index(current_num) == i:
                k += 1

            else:
                nums[i] = '_'
        return k
        # removed_list =  list(set(nums))
        # removed_element_list = (len(nums) - len(removed_list)) * ['_']
        # return len(removed_list), removed_list + removed_element_list

s = Solution()
print(s.removeDuplicates(nums = [1,1,2]))
print(s.removeDuplicates(nums = [4,4,0,0,1,1,1,2,2,3,3,4]))
print(s.removeDuplicates(nums = [1,1,2]))
```

```python
from typing import List

class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        if not nums:
            return 0
        
        # اشاره‌گر برای نگهداری مکان‌های یکتا
        k = 1  # اولین عنصر همیشه یکتا است
        for i in range(1, len(nums)):
            if nums[i] != nums[i - 1]:  # اگر عنصر فعلی با قبلی متفاوت بود
                nums[k] = nums[i]  # قرار دادن عنصر یکتا
                k += 1  # جابجا کردن مکان
        return k

# تست
s = Solution()
print(s.removeDuplicates(nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]))  # خروجی: 5, nums = [0, 1, 2, 3, 4, _, _, _, _, _]
print(s.removeDuplicates(nums = [1, 1, 2]))  # خروجی: 2, nums = [1, 2, _]
```
