```python
from typing import Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def deleteDuplicates(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return None
        
        new_list = ListNode(0)
        new_list.next = head
        current = new_list
        
        
        while current.next and current.next.next :
            if current.next.val == current.next.next.val:
                current.next = current.next.next
            else:
                current = current.next
            
        return new_list.next
                    
        

head1 = None
head2 = ListNode(1)
head3 = ListNode(1, ListNode(2, ListNode(3, ListNode(4))))
head4 = ListNode(1, ListNode(1, ListNode(2, ListNode(3, ListNode(3, ListNode(4))))))


solution = Solution()
test_cases = [ head1, head2,head3, head4]

for i, case in enumerate(test_cases, 1):
    print(f"Test Case {i}:")
    result = solution.deleteDuplicates(case)

    while result:
        print(result.val, end=" → " if result.next else "\n")
        result = result.next
```
