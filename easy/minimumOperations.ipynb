```python
from typing import List
class Solution:
    def minimumOperations(self, nums: List[int]) -> int:
        counter = 0
        while len(nums) != len(set(nums)):
            if len(nums) > 2 :
                nums = nums[3:]
            else:
                nums = []
            counter += 1
        return counter

s = Solution()
print(s.minimumOperations(nums =[1,2,3,4,2,3,3,5,7]))
print(s.minimumOperations(nums = [4,5,6,4,4]))
print(s.minimumOperations(nums = [4,5,6,2,0]))
print(s.minimumOperations(nums = [4,5,4,1,4,3]))
```

```python
from typing import List

class Solution:
    def minimumOperations(self, nums: List[int]) -> int:
        duplicate_count = len(nums) - len(set(nums))
        return (duplicate_count // 3) + (1 if duplicate_count % 3 else 0)

print(s.minimumOperations(nums =[1,2,3,4,2,3,3,5,7]))
print(s.minimumOperations(nums = [4,5,6,4,4]))
print(s.minimumOperations(nums = [4,5,6,2,0]))
print(s.minimumOperations(nums = [4,5,4,1,4,3]))
```
