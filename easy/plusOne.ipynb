```python
from typing import List
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        if digits[-1] != 9:
            digits[-1] +=1
            return digits
            
        if digits.count(9) == len(digits):
            new_digits = [0] * len(digits)
            new_digits.insert(0,1)
            return new_digits

        for i in range(len(digits) - 1, -1, -1):   
            if digits[i] == 9:
                digits[i] = 0
            else:
                digits[i] += 1
                return digits


s = Solution()
print(s.plusOne(digits = [1,2,3]))
print(s.plusOne(digits = [9,9,9,9]))
print(s.plusOne(digits = [4,3,2,1]))
print(s.plusOne(digits = [4,3,9,9]))
```
